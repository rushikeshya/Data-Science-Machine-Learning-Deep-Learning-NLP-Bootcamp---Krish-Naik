# File Operation - Read And Write Files
- File handling is a crusial part of any programming language. Python provides buit-in functions and methods to read from and write files, both text and binary.
- This lesson will cover the basics of file handling, including reding and writing text files and binary files.

```python
with open("example.txt", "r") as file:
    content = file.read()
    print(content)
```

### Read a File line by line

```python
with open("example.txt","r") as file:
    for line in file:
        print(line)
```

```python
with open("example.txt", 'r') as file:
    for line in file:
        print(line.strip())  # strips() removes the newline character
```

```python
## Writing a file (Overwriting)

with open("example.txt", 'w') as file:
    file.write("Written by Rushikesh!\n")
    file.write("This is 2nd Line! \n")
    file.writelines("This is writeline\n")
    file.writelines("Tis is 2nd Writeline\n")
    file.write("This is only write\n")
```

```python
with open("example.txt", 'r') as new_file:
    print(new_file.read())
```

```python
with open("example.txt", 'r') as new_file:
    print(new_file.readline())
```

```python
with open("example.txt", "rb") as binary_file:  # Use "rb" for reading in binary mode
    content = binary_file.read()  # Read the entire content
    print(content)  # This will print the raw bytes

    # If you want to decode it to a string (assuming it's text):
    try:
        decoded_content = content.decode('utf-8')  # Change 'utf-8' to the appropriate encoding if needed
        print(decoded_content)  # Print the decoded string
    except UnicodeDecodeError:
        print("The content is not valid UTF-8 text.")
```

##  Handeling Binary Files

```python
## Writing to a binary file

data = b'\x00\x01\x02\x03\x04'
with open('example.bin', 'wb') as file:
    file.write(data)
```

### Read the content from source text file and write to a destination text file

```python
with open('example.txt','r') as source_file:
    content = source_file.read()

with open("destination.txt", 'w') as destination_file:
    destination_file.write(content)
```

```python
with open('destination.txt', 'r') as result_file:
    print(result_file.read())
```

## Read a text file and count the number of lines, words and characters.


```python
def coun_text_file(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

file_path = 'example.txt'
lines, words, characters = coun_text_file(file_path)
print(f"Lines: {lines}, Words: {words}, Characters: {characters}")
```

### Writing and then reading a file
- i.e. w+ mode: 
    - The w+ mode in Python is used to open a file for both reading and writing.
    - If the file does not exist, it will be created. If the file exists, its content is truncated(i.e., the file is overwritten)

```python
with open('example.txt', 'w+') as file:
    file.write("Hellow worlds\n")
    file.write("This is a new line\n")

    # Move the file cursor to the beginning
    # file.seek()
    
    content = file.read()
    print(content)
```

```python
with open('example.txt', 'w+') as file:
    file.write("Hellow worlds\n")
    file.write("This is a new line\n")

    # Move the file cursor to the beginning
    file.seek(0)
    
    content = file.read()
    print(content)
```
