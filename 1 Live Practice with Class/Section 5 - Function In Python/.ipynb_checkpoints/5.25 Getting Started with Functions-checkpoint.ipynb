### Introduction to Functions

- Definaion: 
    - A Function is a block of code that performs a specific task. Functions help in organizing code, reuseing code, and improving code readability.

```python
# Syntax
def function_name(parameters):
    """Docstring"""
    # Function Body
    return ("expression")
```

```python
## Why Function?
number = 24
if number%2 == 0:
    print("The number is Even")
else:
    print("Number is Odd")
```

```python
def is_even(num):
    if num % 2 == 0:
        print(f"{num} is Even")
    else:
        print(f"{num} is odd")

num = int(input("Enter the Number to check Even or odd: "))
is_even = is_even(num)
```

```python
def even_or_odd(num):
    """This Function find Even or odd"""
    if num % 2 == 0:
        print(f"{num} is Even NUmber")
    else:
        print(f"{num} is odd number.")
```

```python
even_or_odd(-13)
```

```python
# Function with multiple parameter
def add(a,b):
    return a + b
result = add(4,5)
print(result)
```

```python
def inr_to_usd(a, b=85):
    return f"Your Rupees = ${a / b:.2f} USD"
```

```python

result_INR = usd_to_inr(100)
result_USD = inr_to_usd(875)
print(result_INR)
print(result_USD)
```

```python
# Function with default parameter
def usd_to_inr(a, b=85):
    return (f"Your usd = {a*b} INR")

def inr_to_usd(a, b=85):
    return (f"Your Rupees = ${a / b:.2f} USD")

result_INR = usd_to_inr(100)
result_USD = inr_to_usd(875)
print(result_INR)
print(result_USD) 
```
