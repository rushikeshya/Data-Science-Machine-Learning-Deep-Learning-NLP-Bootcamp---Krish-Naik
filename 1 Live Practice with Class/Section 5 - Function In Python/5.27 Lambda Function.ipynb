# Lambda Functions in Python
- Lambda function are small anonymous function defined using the **lambda** keyword. 
- They can have any number of arguments but only one expression. 
- They are commonly used for short operations or as arguments to higher-order functions.

_Note That:_ Anonymous = Function without name

#### Syntax = 
- lambda arguments: expression

```python
def add(a,b):
    return a + b
add(4,5)
```

```python
adds = lambda a,b : a+b
print(type(adds))
print(adds(4,3))
```

```python
# sec 5.27 - 03:00
def even(num):
    if num%2 == 0:
        return True
    else:
        return False
is_even = even(int(input("Enter the number: ")))
print(is_even)
```

```python
# Using lambda
even1 = lambda num : num%2 == 0
even1(int(input("Enter the number: ")))
```

```python
def add_multi(x,y,z):
    return x+y+z
add_multi(4,3,2)
```

```python
add_multi1 = lambda x,y,z: x+y+z
add_multi1(4,3,1)
```

```python
#### map() - Applies a function to all items in a list
number = [1,2,3,4,5,6,7]
def square(num):
    return num ** 2
square(3)
```

```python
squa = lambda x : x**2
squa(3)
```

```python
map(lambda x: x**2, number)
```

```python
list(map(lambda x: x**2, number))
```
