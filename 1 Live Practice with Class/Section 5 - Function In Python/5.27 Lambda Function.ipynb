# Lambda Functions in Python
- Lambda function are small anonymous function defined using the **lambda** keyword. 
- They can have any number of arguments but only one expression. 
- They are commonly used for short operations or as arguments to higher-order functions.

_Note That:_ Anonymous = Function without name

#### Syntax = 
- lambda arguments: expression

```python
def add(a,b):
    return a + b
add(4,5)
```

```python
addition =  lambda x,y: x+y
print(type(addition))
print(type(add))
print(addition(4,5))
```
