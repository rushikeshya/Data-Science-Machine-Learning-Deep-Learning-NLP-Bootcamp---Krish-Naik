## loc and iloc
#### 1. Indexing Method
- loc:

    - Uses label-based indexing.
    - You can access rows and columns by their labels (i.e., the actual names of the index or column).
    - Syntax: df.loc[row_label, column_label]

- iloc:

    - Uses integer-based indexing.
    - You can access rows and columns by their integer position (i.e., the index number).
    - Syntax: df.iloc[row_index, column_index]

#### 2. Inclusivity
    - loc:

        - The end index is inclusive. For example, if you use df.loc[0:1], it will include both rows 0 and 1.
        
    - iloc:
        - The end index is exclusive. For example, if you use df.iloc[0:1], it will include only row 0.

#### 3. Use Cases
    - loc:
        - Useful when you want to access data based on the actual labels of the DataFrame.
        - Example: If you have a DataFrame with a custom index (like names or dates), you would use loc to access
        data by those labels.

    - iloc:
        - Useful when you want to access data based on the position of the rows and columns, regardless of their labels.
        - Example: If you want to access the first row or the first column without caring about what they are named, you would use iloc.

```python
data = {
    'Name':['krish', 'John', 'Jack'],
    'Age':[25, 30, 45],
    'City': ['Pune', 'Mumbai', 'Solapur']
}
df_0 = pd.DataFrame(data)
```

#### loc example - label-based indexing:

```python
df_0
```

```python
df_0.loc[0:1, ["Name", "City"]]  # Access rows 0 to 1 and columns 'Name' and 'City'
```

#### iloc Example -  position-based indexing:

```python
df_0.iloc[0:2,[0,2]]  # Access rows 0 to 1 (exclusive of 2) and columns at index 0 and 2
```

## Accessing a specified element

##### Accessing a specified element using `at` - `row index` and `columns label`

```python
data = {
    'Name':['krish', 'John', 'Jack'],
    'Age':[25, 30, 45],
    'City': ['Pune', 'Mumbai', 'Solapur']
}
df_0 = pd.DataFrame(data)
```

```python
df_0.at[1, 'Age']
```

```python
df_0.at[2, 'Name']
```

```python
df_0
```

##### Accessing a specified element using `iat` - i.e. `row index` and `column index`

```python
df_0.iat[2,2]
```

```python
df_0.iat[2,1]
```
