```python
df_fill = df.fillna(df["Profit"].mean())
df_fill.head()
```

```python
df["New_Amount"] = df["Amount"].apply(lambda x:x*2)
df.head()
```

## Data Aggregatting And Grouping

```python
group_by_Cate = df.groupby('Category')['Profit'].mean()
group_by_Cate.head()
```

```python
# Group by using 2 Categorical Columns
group_by_cate_2 = df.groupby(['Category', 'PaymentMode'])['Quantity'].sum()
group_by_cate_2
```

## Aggregate multiple functions

```python
#16.36
grouped_agg = df.groupby("Category")['Profit'].agg(['mean', 'sum', 'count'])
grouped_agg
```

## Merging and joining DataFrame

```python
# Create sample DataFrames
df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1, 2, 3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value2': [4, 5, 6]})
```

```python
df1
```

```python
df2
```

## Merge DataFrame on the 'Key' Columns

```python
df_merged = pd.merge(df1, df2, on='Key', how='inner')
df_merged
```

```python
df1.merge(df2, on='Key', how='left')
```

```python
df1.merge(df2, on='Key', how='right')
```

```python
df1.merge(df2, on='Key', how='outer')
```
