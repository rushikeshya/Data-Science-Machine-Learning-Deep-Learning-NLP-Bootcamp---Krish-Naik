## Instance Variable and Methods

```python
class Car:
    def __init__(self, color, model, year):
        self.color = color  # instance variable
        self.model = model  # instance variable
        self.year = year    # 
    def expiry_date(self):
        self.extra_date = self.year + 35 
        if self.model == "Honda":
            print(f"Expiry Date of {self.model} is on {self.extra_date}")
        else:
            print(f"Expiry Date of {self.model} is on {self.year + 30}")

# Creating objects (instances) of the Car class
my_car = Car("red", "Toyota", 2020)
your_car = Car("blue", "Honda", 2021)

# Accessing instance variables
print(my_car.color)  # Output: red
print(your_car.model)  # Output: Honda
print(your_car.expiry_date())
```

## Instance Meathods

```python
## Define a class with instance mthods

class Dog:
    # Constructor
    def __init__(self, name, age):
        self.name = name          # Instance Variable
        self.age = age            # Instance Variable
    # Instance Methods
    def bark(self):
        print(f"{self.name} says woof")
# objects
dog1 = Dog("Tommy", 5)
print(dog1.name)
dog1.bark()
```

```python
Dog2 = Dog("Lucy", 3.5)
print(Dog2.age)
```

# Modeling a Bank Accout

```python
### Modeling a Bank Account

## Define a class for bank account
class BankAccount:
    # Constructor
    def __init__(self, owner, balance=0):
        self.owner = owner  # Instance Variables
        self.balance = balance

    # Instance methods
    def deposite(self, amount):
        self.balance += amount
        print(f"{amount} RS deposited! New balance is {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient Found(Money)")
        else:
            self.balance -= amount
            print(f"{amount} RS withdrawn Successfully! and Available Balance is {self.balance}")
    
    def check_balance(self):
        return self.balance
    
## Create an account (Object):
ac1 = BankAccount("Rushikesh")
print(ac1.owner)

# calling instance methods
ac1.deposite(500)
ac1.deposite(200)
ac1.withdraw(70)
ac1.check_balance()
    
```
