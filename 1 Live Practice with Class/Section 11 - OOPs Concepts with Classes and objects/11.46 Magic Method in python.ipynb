- Magic methods in python is also known as dunder methods (double underscore methods), are special methods that start and end with double underscores.
- These methods enable you to define the behavior of objects for built-in operations, such as arithmatic operations, comparisons, and more

### Magic Methods
- Magic methods are predefined methods in python that we can override to change the behavior of our objects. Some common magic methods include:

    - `__init__`: Initializes a new instance of a class.
    - `__str__`: Returns a string representation of an objects.
    - `__repr__`: Returns an official string representation of an object.
    - `__len__`: Returns the length of an object.
    - `__getitem__`: Gets an item from a container.
    - `__setitem__`: Sets an item in a container.

```python
# Basic Methods
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # method overriding
    def __str__(self):
        return f"{self.name}, {self.age} years old"
    
    def __repr__(self):
        return f"Person(name={self.name}, age={self.age})"

person = Person("Rushikesh", 25)
print(person)
print(repr(person))
```

# 11.47 - Operator Overloading in Python
- Operator overloading allows you to define the behavior of operators (+, -, *,)etc for custom objects. We achive this by overriding  specific magic methods in our class.

#### Common Operators Overloading Magic Methods

`__add__(self, other)`: Adds two objects using the  + operators.

`__sub__(self, other)`: Substracting two objects using the - operators.

`__mul__(self, other)`: Multiplies two objects using the * operators.

`__truediv__(self, other)`: Divides two objects using the / operators.

`__eq__(self, other)`: Check if two objects are equal using the == operator.

`__lt__(self, other)`: Check if one object is less than another using the < operator.

`__gt__(self, other)`: Check if one object is Greter than another using the > operator.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    
    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)
    
    def __mul__(self, other):
        return Vector(self.x * other.x, self.y * other.y)
    
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

# Create objects of the Vector Class
v1 = Vector(2,3)
v2 = Vector(4,5)

print(v1 + v2)
print(v1 - v2)
print(v1 * v2)
```
