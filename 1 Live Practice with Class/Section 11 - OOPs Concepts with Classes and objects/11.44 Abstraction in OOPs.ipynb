# Abstraction in OOPs
- Abstract means **hiding the complex details** and showing only the essential features of an object

- Abstraction is the concept of hiding the complex implementation details and sowing only the necessary features of an object.
- This helps in reducing programming complexity and effort.

```python
from abc import ABC, abstractmethod
# Abstract base class
class Vehicle(ABC):
    def drive(self):
        print(f"The vehicle is used for diving!")
    
    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print("Car Engine Started")

# Demonstarte Operation of Abstract
def operate_vehicle(vehicle):
    vehicle.start_engine()
    vehicle.drive()

car = Car()
operate_vehicle(car)
```

# 🔧 Real-Life Problem Solving Example (Python Code):
Problem: Create a system for sending notifications (Email & SMS)
We’ll use abstraction to define a general "Notification" class and let child classes handle specific details.

```python
from abc import ABC, abstractmethod

# Abstract Base Class
class Notification(ABC):
    @abstractmethod
    def send(self, message):
        pass

# Derived class for Email
class EmailNotification(Notification):
    def send(self, message):
        print(f"Sending Email: {message}")
    
# Derived class for SMS
class SMSNotification(Notification):
    def send(self, message):
        print(f"Sending SMS: {message}")

# Function that uses abstraction
def notify_user(notification_type, message):
    notification_type.send(message)

# Objects
email = EmailNotification()
sms = SMSNotification()
notify_user(email, "Your order has been shipped")
notify_user(sms, "Your otp is 12345")
```

# 🧩 Problem Statement (Abstraction Practice):
You are designing a Payment System.
Create an abstract class called `PaymentMethod` with an abstract method pay(amount).
Then, implement two classes:

- `CreditCardPayment` – simulate payment using a credit card.

- `UPIPayment` – simulate payment using a UPI app (like Google Pay, PhonePe).

Use a function `process_payment(payment_method, amount)` to handle the payment without knowing the internal details.

```python
from abc import ABC, abstractmethod

# Abstract Base class
class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

# 1st derived class:    
class CreditCardPayment(PaymentMethod):
    def pay(self, amount):
        print(f"Thank you for payment with Credit Card! {amount} credited")
    
# 2nd derived class
class UPIPayment(PaymentMethod):
    def pay(self, amount):
        print(f"Thank you for choosing UPIPayment method! {amount} has been send")

# Function  that uses abstract
def  process_payment(payment_method, amount):
    return payment_method.pay(amount)

# objects

credit = CreditCardPayment()
upi = UPIPayment()

process_payment(upi,455)
process_payment(credit, 500)
```
