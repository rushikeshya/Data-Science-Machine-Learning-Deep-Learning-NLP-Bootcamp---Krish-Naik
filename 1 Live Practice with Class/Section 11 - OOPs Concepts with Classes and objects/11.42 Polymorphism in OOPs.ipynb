# Polymorphism in OOPs
- Polymorphism is a core concept in Object-Oriented Programming (oops) that allows objects of different classes to be treated as objects of a common superclass.
- `It provides a way to perform a single action in different forms.`
- `Polymorphism is typically through method overriding and interface` 

## Method Ovrriding - 
- Method overriding allows a child class to provide a specific implementation of a method that is alredy defined in its parent class

```python
class Animal:
    def speak(self):
        return f"Animal sounds!"
    
## Derived Class 1st
class Dog(Animal):
    def speak(self):
        return "Woof!"

## Derived Class 2nd
class Cat(Animal):
    def speak(self):
        return "Meow!"
    
# Objects Calling
dog = Dog()
cat = Cat()
print(dog.speak())
print(cat.speak())

animal = Animal()
print(animal.speak())
```

```python
class Animal:
    def speak(self):
        return f"Animal sounds!"
    
## Derived Class 1st
class Dog(Animal):
    def speak(self):
        return "Woof!"

## Derived Class 2nd
class Cat(Animal):
    def speak(self):
        return "Meow!"

# Function that demonstrates polymorphism
def animal_speak(animal):
    print(animal.speak())



# Objects Calling
dog = Dog()
cat = Cat()
print(dog.speak())
print(cat.speak())

# polymorphism
animal_speak(dog)
```

## Polymorphism with functions and methods

```python
## Base Class
class Shape:
    def area(self):
        return "The area of the figure"

## Derived Class 1st:
class Reactangle(Shape):
    # Constructor
    def __init__(self, width, height):
        self.width = width
        self.height = height

    # Redifining the function that is already defined in Parent class `Shape` 
    def area(self):
        return self.width * self.height

## Derived Class 2:
class Circle(Shape):
    # Constructor
    def __init__(self, radius):
        self.radius = radius
    
    # Redifining the function that is already defined in Parent class `Shape` 
    def area(self):
        return 3.14 * self.radius * self.radius
    
## Function
def print_area(shape):
    print(f"The Area is {shape.area()}") 

# Objects
rectangle = Reactangle(4,5)
circle = Circle(40)

print_area(rectangle)
print_area(circle)
```

# Polymorphism with Interfaces (Abstract Base Class)
- Abstract Base Classess (ABSc) are used to define common methods for a group of related objects. They can enforce that derived class implement particular methods, promoting consistency across different implementations
- from `abc` import `ABC`, `abstractmethod`

```python
from abc import ABC, abstractmethod

# Define an abstract class
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

# Derived Class 1
class Car(Vehicle):
    def start_engine(self):
        return f" Car engine stared"

# Derived Class 2
class Motorcycle(Vehicle):
    def start_engine(self):
        return "Motorcycle engine start"

## Function that demonstrates polymorphism
def start_vehicle(vehicle):
    print(vehicle.start_engine())


# Objects
car = Car()
motorcycle = Motorcycle()

print(start_vehicle(car))
print(start_vehicle(motorcycle))
```
