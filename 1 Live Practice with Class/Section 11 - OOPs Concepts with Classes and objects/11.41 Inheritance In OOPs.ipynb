# Inheritance In OOPs
- Inheritance is a fundamental concept in Object-Oriented Programming (OOP) that allows a class to inherit attributes and methods from another class. This lesson covers single inheritance and multiple inheritance, demonstrating how to create and use them in Python

```python
class Car:
    def __init__(self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
    
    def drive(self):
        print(f"The person will drive the {self.enginetype} car.")

# Test
car1 = Car(4,5,"petrol")
car1.drive()
```

```python
class Tesla(Car):
    def __init__(self, windows, doors, enginetype, battery_size):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
        self.battery_size = battery_size
    def check_features(self):
        if self.enginetype == "electrical":
            return f"The Car is Electrical Car and it engine is {self.enginetype} and battery = {self.battery_size}"
        else:
            pass
# Test
car2 = Tesla(4, 5, "electrical", "400kw")
print(car2.check_features())
car2.drive()
            
```

## Single Inheritance

```python
class Tesla(Car):
    def __init__(self, windows, doors, enginetype, battery_size, is_selfdriving):
        super().__init__(windows, doors, enginetype)   # Inheritence from parent class
        self.battery_size = battery_size
        self.is_selfdriving = is_selfdriving

    def selfdriving(self):
        print(f"Tesla support self driving : {self.is_selfdriving}")

    def check_features(self):
        if self.enginetype == "electrical":
            return f"The Car is Electrical Car and it engine is {self.enginetype} and battery = {self.battery_size}"
        else:
            pass
# Test
car2 = Tesla(4, 5, "electrical", "400kw", True)
print(car2.check_features())
car2.drive()

car3 = Tesla(4,5,"electrical", "550kw", True)
print(car3.is_selfdriving)
car3.selfdriving()
```

## Multiple Inheritance
- When a class inherits from more than one base class.

```python
# Base class 1
class Animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        print("Subclass must implement this method")

# Base class 2
class pet:
    def __init__(self, owner):
        self.owner = owner

## Derived Class
class Dog(Animal, pet):
    def __init__(self, name, owner):
        # super().__init__(name, owner)   #this is for only single inheriance
        Animal.__init__(self, name)
        pet.__init__(self, owner)

    def speak(self):
        return f"{self.name} say woof!"

    
d1 = Dog("Puppy","Some")
print(d1.owner)
print(d1.speak())
```
