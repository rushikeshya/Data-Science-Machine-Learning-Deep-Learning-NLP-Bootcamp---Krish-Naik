# Generators:
- Generators are the simple way to create iterators. 
- They use the yield keyword to produce a series of values lazily, which means they generate values on the fly and do not store them in memory.

```python
def square(n):
    for i in range(3):
        return i**2
square(3)
```

```python
def square(n):
    for i in range(3):
        yield i**2    # yield keyword create local variable and save the value
square(3)
```

```python
for i in square(3):
    print(i)
```

```python
a = square(3)
a
```

### Pratical Example: Reading Large File:
- Generators are particularly useful for reding large files beacuase they allow you to process one line at a time without loading the file into memory.

```python
def read_large_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line
```

```python
file_path = "large_file.txt"
for line in read_large_file(file_path):
    print(line.strip())
```
