# Iterators In Python
- Iterators are advanced Python concepts that allow for efficient looping and memory management.
- Iterators provide a way to access elements of a collection sequentially without exposing the underlying structure.

```python
my_list = [1,2,3,4,5,6]
for i in my_list:
    print(i)
```

```python
type(my_list)
```

## Iterator

```python
iterator = iter(my_list)
print(iterator)
```

```python
print(type(iterator))
```

### Iterate through all the element we use next() built-in func.

```python
# Lazy Loading
next(iterator)
```

```python
iterator = iter(my_list)
```

```python
try:
    print(next(iterator))
except StopIteration as e:
    print(e)
    print("There are no elements in the iterators! ")
```

```python
my_string = "Hello world"
string_iterator = iter(my_string)

print(next(string_iterator))
print(next(string_iterator))
```
