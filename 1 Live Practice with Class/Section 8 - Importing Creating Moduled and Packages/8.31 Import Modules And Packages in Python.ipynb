# 8.31 Import Modules And Packages in Python

- In Python, modules and packages help organize and reuse code. Here's a comprehensive guide on how to import them.

```python
import math
math.sqrt(16)
```

```python
from math import sqrt, pi
pie = pi
pie
```

```python
import numpy as np
np.array([1,2,3,4,5])
```

```python
from math import *
sqrt(36)
```

```python
from package.rush_maths import adding
adding(4,6)
```

```python
from package import rush_maths
rush_maths.adding(7,4)
```
