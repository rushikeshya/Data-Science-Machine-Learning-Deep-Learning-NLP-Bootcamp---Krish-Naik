```python
import os

file_path = "new_file.txt"  # Specify the desired file path

try:
    # Open the file in 'x' mode (exclusive creation)
    with open(file_path, 'x') as file:
        # Optionally, write content to the file
        file.write("This is a new file.")
    print(f"File '{file_path}' created successfully.")
except FileExistsError:
    print(f"File '{file_path}' already exists.")
except Exception as e:
    print(f"An error occurred: {e}")
```

## CSV

```python
import csv

with open('example.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name', 'age'])
    writer.writerow(['Rushi', 24])
with open('example.csv', mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

## Regular expression - re

 - the built-in module for working with regular expressions (regex), which are powerful tools for pattern matching and text manipulation. The re module provides functions to search, match, and replace substrings within text based on specified patterns. 

**Here's a more detailed explanation:**

**What is regex?**

- Regular expressions are sequences of characters that define a search pattern. 

- They're used to check if a string contains a specific pattern, or to find, replace, or manipulate text based on patterns.

**Why use regex?**

- Data Validation: Regex can be used to validate the format of data, such as email addresses, phone numbers, or passwords. 

- Text Parsing: They are helpful for parsing text data files, finding, replacing, or deleting specific strings. 

- String Manipulation: Regex allows for efficient string manipulation tasks, such as extracting specific parts of a string or replacing parts with other strings. 

How to use the re module 
Import the module: import re 

**Common functions:**
 
**re.search():** Searches for a match anywhere in the string. 
**re.match():** Searches for a match only at the beginning of the string. 
**re.findall():** Returns a list containing all matches. 
**re.sub():** Replaces one or many matches with a string. 
**re.compile():** Compiles a regular expression pattern into a regular expression object for efficient reuse. 

```python
import re

pattern = r'\d+'  # d for digits
text = "There are 123 apples 456"
match = re.search(pattern, text)
print(match.group())
```
