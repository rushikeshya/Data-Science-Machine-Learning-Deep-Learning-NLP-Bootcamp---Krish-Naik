Exercise 1: Create a class Greeter with a method greet(name) that prints a greeting for the provided name.

```python
class Greeter:
    def greet(self,name):
        print(f"Hello, Good Morning! {name}")
#objects
person_1 = Greeter()
person_1.greet("Rushikesh")
```

Exercise 2: Develop a class Calculator with methods to add and subtract two numbers.


```python
class Calculator:
    def add(self, a, b):
        return a+b
    def sub(self, a,b):
        return a-b
c = Calculator()
print(c.add(4,5))
print(c.sub(4,5))
```

Exercise 3: Build a class Employee with multiple constructors that can initialize an employee object in different ways.





```python
class Employee:
    def __init__(self, name, id=None, department=None):
        self.name = name
        self.id = id
        self.department = department
    def display_details(self):
        print(f"Name: {self.name}")
        if self.id:
            print(f"ID: {self.id}")
        if self.department:
            print(f"Department: {self.department}")

# Objects
e1 = Employee("Rushi",None,"Data Science")
e1.display_details()
```

Exercise 5: Create a class MaxFinder that identifies the largest number in a list


```python
class MaxFinder:
    def __init__(self, lst):
        self.lst = lst
    
    def find_max(self):
        if not self.lst:
            return "List is Empty"
        return max(self.lst)

finder = MaxFinder([1,5,7,11,20,19])
print("The largest muber is: ",finder.find_max())
```

Exercise 6: Design a Rectangle class with default attributes for length and width set to 1. Include methods to set these attributes and calculate the area.

```python
class Reactangle:
    def __init__(self, length=1, width=1):
        self.length = length
        self.width = width
    
    def set_dimensions(self, length, width):
        self.length = length
        self.width = width
    
    def area(self):
        return self.length * self.width

# Objcts
rect = Reactangle()
print("Default area: ", rect.area())
rect.set_dimensions(4, 5)
print("Updates area: ", rect.area())
```
