# Bitwise Operators
| Operator | Name             | Meaning                                                             | Example                          |
|----------|------------------|---------------------------------------------------------------------|----------------------------------|
| `&`      | AND              | 1 if **both bits** are 1, else 0                                    | `5 & 3` → `0101 & 0011 = 0001 (1)` |
| `|`      | OR               | 1 if **either bit** is 1, else 0                                    | `5 | 3` → `0101 | 0011 = 0111 (7)` |
| `^`      | XOR              | 1 if **bits are different**, 0 if same                              | `5 ^ 3` → `0101 ^ 0011 = 0110 (6)` |
| `~`      | NOT (Complement) | Flip all bits (0 → 1, 1 → 0), works on **whole number** (2's comp.) | `~5` → `-(5 + 1) = -6`            |
| `<<`     | Left Shift       | Shift bits left (equivalent to multiplying by 2)                   | `5 << 1` → `0101 << 1 = 1010 (10)` |
| `>>`     | Right Shift      | Shift bits right (equivalent to dividing by 2)                     | `5 >> 1` → `0101 >> 1 = 0010 (2)`  |

# ~	NOT (Complement)
- Flip all bits (0 becomes 1, 1 becomes 0), works on whole number
- ~5 = -6 (2’s complement logic)

````python
#0101 #bin(5)

#1010 #bin(6)
````

````python
bin(5)
````

````python
n = 5
result = ~n
print("~{} = {}".format(n, result))   # Output: ~5 = -6

# Or show binary:
print("Binary of 5      :", format(n, '08b'))          # 00000101
print("Binary of ~5     :", format(~n & 0xFF, '08b'))  # 11111010 (8-bit)
````
