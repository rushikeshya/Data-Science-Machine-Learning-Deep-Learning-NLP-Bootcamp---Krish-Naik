```python
ages = [11,12,14,18,24,26,30,35,36,37,40,41,42,43,50]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,4))
plt.hist(ages, bins=5, histtype='bar', density=True, label='Histogram')

sns.kdeplot(ages, color='red',label='KDE',linewidth=2)

plt.xlabel("Ages")
plt.ylabel("Probability Density")
plt.title("Probability Distribution of Ages")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Sample data for ages
ages = np.random.normal(loc=30, scale=10, size=1000)  # Example data

plt.figure(figsize=(8, 4))

# Plot histogram with density=True
plt.hist(ages, bins=30, density=True, histtype='step', label='Histogram')

# Overlay a kernel density estimate (KDE)
sns.kdeplot(ages, color='red', label='KDE', linewidth=2)

plt.xlabel("Ages")
plt.ylabel("Probability Density")
plt.title("Probability Distribution of Ages")
plt.legend()
plt.show()
```
