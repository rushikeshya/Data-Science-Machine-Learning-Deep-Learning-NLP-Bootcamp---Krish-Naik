| Big-O        | Name          | Analogy / Example                                      |
| ------------ | ------------- | ------------------------------------------------------ |
| `O(1)`       | Constant Time | Pick the first item from a box                         |
| `O(log n)`   | Logarithmic   | Binary search in a sorted list                         |
| `O(n)`       | Linear Time   | Look at every item once                                |
| `O(n log n)` | Linearithmic  | Efficient sorting like Merge Sort or Quick Sort        |
| `O(n^2)`     | Quadratic     | Nested loops - compare every pair                      |
| `O(2^n)`     | Exponential   | Subset generation (e.g. recursive Fibonacci)           |
| `O(n!)`      | Factorial     | All possible arrangements (e.g. permutations of items) |

### O(n) — one loop

```python
n = 2
for i in range(n):
    print(i)
```

### O(n²) — nested loop

```python
n = 2
for i in range(n):
    for j in range(n):
        print(i, j)
```

### # O(log n) — binary search

```python
def binary_search(arr, x):
    l, r = 0, len(arr)-1
    while l <= r:
        mid = (l + r) // 2
        if arr[mid] == x:
            return mid
        elif arr[mid] < x:
            l = mid + 1
        else:
            r = mid - 1
```

```python
binary_search([1,2,3,4,5], 4)
```
