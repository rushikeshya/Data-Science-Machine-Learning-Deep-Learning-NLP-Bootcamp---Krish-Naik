### Assignment 1: Polymorphism with Methods

Create a base class named `Shape` with a method `area`. Create two derived classes `Circle` and `Square` that override the `area` method. Create a list of `Shape` objects and call the `area` method on each object to demonstrate polymorphism.

```python
import math

# Base class 1
class Shape:
    def area(self):
        pass

# Derived class 1
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return math.pi * self.radius ** 2
    
# Derived class 2
class Square(Shape):
    def __init__(self, side):
        self.side = side
    
    def area(self):
        return self.side ** 2



# Object
shapes = [Circle(5), Square(4)]
for shape in shapes:
    print(shape.area())   # Polymorphism implementation
```

### Assignment 3: Abstract Base Class with Abstract Methods

Create an abstract base class named `Vehicle` with an abstract method `start_engine`. Create derived classes `Car` and `Bike` that implement the `start_engine` method. Create objects of the derived classes and call the `start_engine` method.

```python
from abc import ABC, abstractmethod

# Base Abstract Class
class Vehicle(ABC):
    @abstractmethod          # Abstract Method
    def start_engine(self):
        return "Engine Stared....!"
    
# Derived class 1
class Car(Vehicle):
    def start_engine(self):
        print("Car Engine Started...!")

# Derived class 2
class Bike(Vehicle):
    def start_engine(self):
        print("Bike Engine Startes......!")

# Objects
car = Car()
bike = Bike()
car.start_engine()
bike.start_engine()
```

### Assignment 5: Encapsulation with Private Attributes

Create a class named `BankAccount` with private attributes `account_number` and `balance`. Add methods to deposit and withdraw money, and to check the balance. Ensure that the balance cannot be accessed directly.

```python
class BankAccount:
    # Constructors
    def __init__(self,account_number, balance = 0):
        self.__account_number = account_number  # Private Attribuites
        self.__balance = balance                # Private Attribuites
    
    # Deposit Methods
    def deposit(self, amount):
        self.__balance += amount
        print(f"{amount} Credited on your a/c {self.__account_number}")
    
    # Withdrawl method
    def withdraw(self, amount):
        if amount > self.__balance:
            print("Amount is Insufficient! ")
        else:
            self.__balance -= amount
            print(f"Your current balance is {self.__balance}, for a/c no. {self.__account_number}")
    
    # Check balance method
    def check_balance(self):
        return self.__balance


# Objects
ac = BankAccount(12345, 500)
ac.deposit(400)
print(ac.check_balance())
ac.withdraw(470)
print(ac.check_balance())
```

### Assignment 6: Encapsulation with Property Decorators

In the `BankAccount` class, use property decorators to get and set the `balance` attribute. Ensure that the balance cannot be set to a negative value.

```python
class BankAccount:
    # Constructors
    def __init__(self,account_number, balance = 0):
        self.__account_number = account_number  # Private Attribuites
        self.__balance = balance                # Private Attribuites
    
    # Property Decorators
    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, amount):
        if amount < 0:
            print("Amount should be greater than ZERO (0)")
        else:
            self.__balance = amount

    # Deposit Methods
    def deposit(self, amount):
        self.__balance += amount
        print(f"{amount} Credited on your a/c {self.__account_number}")
    
    # Withdrawl method
    def withdraw(self, amount):
        if amount > self.__balance:
            print("Amount is Insufficient! ")
        else:
            self.__balance -= amount
            print(f"Your current balance is {self.__balance}, for a/c no. {self.__account_number}")
    


# Objects
ac = BankAccount(12345, 500)
ac.deposit(400)

ac.withdraw(470)
ac.balance = -500
```

### Assignment 7: Combining Encapsulation and Inheritance

Create a base class named `Person` with private attributes `name` and `age`. Add methods to get and set these attributes. Create a derived class named `Student` that adds an attribute `student_id`. Create an object of the `Student` class and test the encapsulation.

```python
from abc import ABC, abstractmethod

# Abstract Base Class
class Employee(ABC):     # This is Implementation class for Abstract method
    @abstractmethod
    def calculate_salary(self):
        print("Sample Salary")

# Derived class 1
class FullTimeEmployee(Employee):
    def __init__(self, salary):
        self.salary = salary

    def calculate_salary(self):
        return self.salary
    
# Derived class 2
class PartTimeEmployee(Employee):
    def __init__(self, hourly_rate, hours_worked):
        self.hourly_rate = hourly_rate
        self.hours_worked = hours_worked
    
    def calculate_salary(self):
        return self.hourly_rate * self.hours_worked

full_time_emp = FullTimeEmployee(35000)
part_time_emp = PartTimeEmployee(500,40)
print(full_time_emp.calculate_salary())
print(part_time_emp.calculate_salary())
# emp = Employee()
# emp.calculate_salary()   # TypeError: Can't instantiate abstract class Employee without an implementation for abstract method 'calculate_salary'
```

### Assignment 10: Encapsulation in Data Classes

Create a data class named `Product` with private attributes `product_id`, `name`, and `price`. Add methods to get and set these attributes. Ensure that the price cannot be set to a negative value.

```python
class Product:
    def __init__(self, product_id, name, price):
        self.__product_id = product_id
        self.__name = name
        self.__price = price
    
    # getter method for product_id
    def get_product_id(self):
        return self.__product_id
    # setter method for product_id
    def set_product_id(self, product_id):
        self.__product_id = product_id
    
    # getter method for name
    def get_name(self):
        return self.__name
    # setter method for name
    def set_name(self, name):
        self.__name = name

    # getter method for price
    def get_price(self):
        return self.__price
    # setter method for price
    def set_price(self, price):
        if price < 0:
            print("Price Cannot be Negative")
        else:
            self.__price = price



# Objects
product = Product(45, "Table Fan", 2140)
print(product.get_product_id(), product.get_name(), product.get_price())
product.set_product_id(46)
product.set_name("Ceiling Fan")
product.set_price(3025)
print(product.get_product_id(), product.get_name(), product.get_price())
product.set_price(3025)
product.set_price(-550)
```

### Assignment 11: Polymorphism with Operator Overloading

Create a class named `Vector` with attributes `x` and `y`. Overload the `+` operator to add two `Vector` objects. Create objects of the class and test the operator overloading.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    
    def __str__(self):
        return f"vector({self.x}, {self.y})"
    
# Objects
v1 = Vector(2,3)
v2 = Vector(4,5)
v3 = v1 + v2
print(v3)
```

### Assignment 12: Abstract Properties

Create an abstract base class named `Appliance` with an abstract property `power`. Create two derived classes `WashingMachine` and `Refrigerator` that implement the `power` property. Create objects of the derived classes and access the `power` property.

```python
from abc import ABC, abstractmethod
class Appliance(ABC):
    @property     # to make sure it behaves like an attribute (not a method)
    @abstractmethod  # to make sure subclass must override it
    def power(self):
        pass
class WashingMachine(Appliance):
    @property
    def power(self):
        return "500W"
class Refrigerator(Appliance):
    @property
    def power(self):
        return "150W"

# Objects
wm = WashingMachine()
ref = Refrigerator()
print(wm.power)  # @property used for method behave like attribute/variable/instances so insted of wm.power() here used wm.power
print(ref.power)
```

### Assignment 14: Polymorphism with Multiple Inheritance

Create a class named `Flyer` with a method `fly`. Create a class named `Swimmer` with a method `swim`. Create a class named `Superhero` that inherits from both `Flyer` and `Swimmer` and overrides both methods. Create an object of the `Superhero` class and call both methods.

```python
class Flyer:
    def fly(self):
        print("Flying...!")
class Swimmer:
    def swim(self):       
        print("Swimming...!")

class Superhero(Flyer,Swimmer):
    def fly(self):   # Method overriding (i.e. Polymorphism)
        print("Superhero flying...!")
    @property
    def swim(self):
        print("Superhero swimming...!")

# objects
superhero = Superhero()
superhero.fly()
superhero.swim   # Using `@property` our swim method behave like attributes/instances/variables
```

### Assignment 15: Abstract Methods and Multiple Inheritance

Create an abstract base class named `Worker` with an abstract method `work`. Create two derived classes `Engineer` and `Doctor` that implement the `work` method. Create another derived class `Scientist` that inherits from both `Engineer` and `Doctor`. Create an object of the `Scientist` class and call the `work` method.

```python
from abc import ABC, abstractmethod
# Abstract Base Class
class Worker(ABC):
    def work(self):
        pass
# derived class 1
class Engineer(Worker):
    def work(self):
        print("Engineer Working...!")

# derived class 2
class Doctor(Worker):
    def work(self):
        print("Doctor Working....!")

# derived class 3
class Scientist(Engineer, Doctor):
    def work(self):
        Engineer.work(self)
        Doctor.work(self)

# Objects
scientist = Scientist()
scientist.work()
```
